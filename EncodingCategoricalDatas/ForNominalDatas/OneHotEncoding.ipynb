OneHotEncoding using Sklearn

```python
from sklearn.model_selection import train_test_split as tts
X_train, X_test, y_train, y_test = tts(df.iloc[:,0:4], df.iloc[:,-1], test_size=0.2, random_state=2)
```

```python
X_test.head()
```

```python
from sklearn.preprocessing import OneHotEncoder as OHE
```

```python
ohe = OHE()
```

```python
X_train_new = ohe.fit_transform(X_train[['fuel','owner']]).toarray()
X_train_new
```

```python
X_test_new = ohe.transform(X_test[['fuel','owner']]).toarray()
X_test_new
```

```python
X_train[['brand','km_driven']].values
```

```python
np.hstack((X_train[['brand','km_driven']].values,X_train_new)).shape
```

Using drop, sparse and dtype function

```python
ohe_new = OHE(drop='first', sparse_output=False, dtype=np.int32) #drops the first column and directly converts it into an array 
```

```python
X_train_neww = ohe_new.fit_transform(X_train[['fuel','owner']])
X_train_neww
```

```python
X_test_neww = ohe_new.transform(X_test[['fuel','owner']])
X_test_neww
```

```python
np.hstack((X_train[['brand','km_driven']].values,X_train_neww)).shape
```
